```python
path = "/run/media/mahdi/New Volume2/Claude/predictive-maintenance-platform/data/raw/Dataset_01/Baseband_02/Dataset_01_LTE_2100.csv"

empty_zero = 0
empty_nonzero = 0
filled_zero = 0
filled_nonzero = 0
total = 0
empty_missing =0
filled_missing = 0
with open(path,newline="")as f:
    for row in csv.DictReader(f):
        total += 1

        rb = row["4G RB utilization"]
        vol = row["4G data volume DL"]
        rb_is_empty = (rb == "") 
        
        vol_is_missing = (vol == "")
        
        if rb_is_empty:
            if vol_is_missing:
                empty_missing += 1
            elif float(vol) == 0:
                empty_zero += 1
            else:
                empty_nonzero += 1    
        else:
            if vol_is_missing:
                filled_missing +=1
            elif float(vol) == 0:
                filled_zero += 1   
            else:
                filled_nonzero += 1  


print(empty_zero, empty_nonzero,empty_missing, filled_zero,filled_missing, filled_nonzero, total)
print(empty_zero + empty_nonzero + empty_missing +filled_zero + filled_nonzero + filled_missing == total)
```
